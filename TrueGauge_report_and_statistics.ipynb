# TrueGauge — Report and Statistics Compendium

### Every real-world figure cited in *"TrueGauge: A Retrofit-Viable, Sensorless-First Vehicle Energy Integrity Indicator..."*, in one place

**What this notebook is.** The paper cites dozens of real statistics — country enforcement actions, market sizing, a technical accuracy benchmark, EV degradation figures, financial-harm cases. This notebook pulls every one of them into structured tables and charts, each traceable back to its exact in-paper citation number (`[1]`-`[50]`), so you (or a reviewer) can audit the evidentiary base of the paper in one pass without re-reading all 10 pages.

**What this notebook is NOT.** It contains no synthetic or simulated data — that is the companion notebook (`TrueGauge_EII_notebook.ipynb`), which implements the formal model (Eq. 1-7) on synthetic data. Every number here is a real figure the paper cites from a named regulator, tax authority, court, market-research report, or peer-reviewed source. Where the paper itself flags a figure as an estimate (e.g., market forecasts, fiscal-loss estimates), that caveat is preserved here rather than smoothed over.

## 5. Full statistic-to-citation index

Every numeric claim used anywhere in the paper, in one lookup table — the fastest way
to trace any figure in the PDF back to exactly where it came from.

```python
all_stats_index = pd.concat([
    countries[["country", "finding", "citation"]].rename(columns={"country": "item", "finding": "statistic"}),
    technical_benchmarks[["metric", "value", "citation"]].rename(columns={"metric": "item", "value": "statistic"}),
    market_stats.assign(statistic=lambda d: "$" + d["price_usd"].astype(str), citation="[37]-[39]")[["item", "statistic", "citation"]],
    harm_cases[["case", "figure", "citation"]].rename(columns={"case": "item", "figure": "statistic"}),
], ignore_index=True)

print(f"Total indexed statistics: {len(all_stats_index)}")
all_stats_index
```
